```python
import subprocess, sys
subprocess.check_call([sys.executable, '-m', 'pip', 'install', '-q', 'xgboost', 'lightgbm', 'scipy'])

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.gridspec as gridspec
import seaborn as sns
import warnings
import os
from scipy.stats import gaussian_kde
from sklearn.model_selection import train_test_split, learning_curve
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LinearRegression, Ridge
from sklearn.ensemble import RandomForestRegressor
from sklearn.dummy import DummyRegressor
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
import xgboost as xgb
import lightgbm as lgb

warnings.filterwarnings('ignore')
os.makedirs('figures', exist_ok=True)
print('All libraries loaded successfully.')
```

```python
df = pd.read_csv('dataset.csv')
df.drop(columns=['Unnamed: 0'], inplace=True, errors='ignore')
df.dropna(inplace=True)

# De-duplicate: keep first occurrence per track_id
# Note: 16,299 track IDs appear in multiple genres; first occurrence is retained
# as genre assignment is not a modelling feature and multi-genre tracks are evenly
# distributed across the dataset.
df_unique = df.drop_duplicates(subset='track_id').copy()
df_unique['explicit']   = df_unique['explicit'].astype(int)   # True→1, False→0
df_unique['duration_s'] = df_unique['duration_ms'] / 1000

# FIX 1: Remove physically invalid rows
# time_signature values 0 and 1 are data artefacts (no song has 0 or 1 beat per bar)
df_unique = df_unique[df_unique['time_signature'].isin([3, 4, 5])].copy()
# tempo == 0 BPM is impossible
df_unique = df_unique[df_unique['tempo'] > 0].copy()

# FIX 2: One-hot encode 'key' (nominal, not ordinal — 12 pitch classes)
# Treating key as a raw integer misleads linear models (key 11 != key 0 + 11 units)
key_dummies = pd.get_dummies(df_unique['key'], prefix='key', drop_first=True)
df_unique   = pd.concat([df_unique.reset_index(drop=True),
                          key_dummies.reset_index(drop=True)], axis=1)
KEY_DUMMIES = [c for c in df_unique.columns if c.startswith('key_')]

FEATURE_COLS = [
    'danceability', 'energy', 'loudness', 'mode',
    'speechiness', 'acousticness', 'instrumentalness',
    'liveness', 'valence', 'tempo', 'time_signature',
    'duration_s', 'explicit'
] + KEY_DUMMIES    # 13 original + 11 key dummies = 24 features total

TARGET_COL = 'popularity'

X = df_unique[FEATURE_COLS].copy()
y = df_unique[TARGET_COL].copy()

# 70 / 15 / 15 split
X_temp, X_test, y_temp, y_test = train_test_split(X, y, test_size=0.15, random_state=42)
X_train, X_val, y_train, y_val = train_test_split(X_temp, y_temp, test_size=0.1765, random_state=42)

# Scale for linear models only — fit on train, transform val/test
scaler      = StandardScaler()
X_train_sc  = scaler.fit_transform(X_train)
X_val_sc    = scaler.transform(X_val)
X_test_sc   = scaler.transform(X_test)

print(f'Dataset : {df_unique.shape[0]:,} unique tracks x {len(FEATURE_COLS)} features')
print(f'Train   : {X_train.shape[0]:,}  |  Val: {X_val.shape[0]:,}  |  Test: {X_test.shape[0]:,}')
print(f'Target  : mean={y.mean():.2f}, std={y.std():.2f}, min={y.min()}, max={y.max()}')
print(f'Tracks with popularity=0 : {(y==0).sum():,} ({100*(y==0).mean():.1f}%)  — may include unlisted/archived tracks')
```

```python
print('Training 6 models...')

# 1. Baseline
dummy = DummyRegressor(strategy='mean')
dummy.fit(X_train_sc, y_train)
y_pred_dummy = dummy.predict(X_test_sc)
print('  [1/6] Baseline (Mean)       done')

# 2. Linear Regression
lr = LinearRegression()
lr.fit(X_train_sc, y_train)
y_pred_lr = lr.predict(X_test_sc)
print('  [2/6] Linear Regression     done')

# 3. Ridge Regression
ridge = Ridge(alpha=10.0, random_state=42)
ridge.fit(X_train_sc, y_train)
y_pred_ridge = ridge.predict(X_test_sc)
print('  [3/6] Ridge Regression      done')

# 4. Random Forest
rf = RandomForestRegressor(n_estimators=200, max_depth=15,
                           min_samples_leaf=4, n_jobs=-1, random_state=42)
rf.fit(X_train, y_train)
y_pred_rf = rf.predict(X_test)
print('  [4/6] Random Forest         done')

# 5. XGBoost
xgb_model = xgb.XGBRegressor(
    n_estimators=300, learning_rate=0.05, max_depth=6,
    subsample=0.8, colsample_bytree=0.8, random_state=42,
    verbosity=0, eval_metric='rmse'
)
xgb_model.fit(X_train, y_train, eval_set=[(X_val, y_val)], verbose=False)
y_pred_xgb = xgb_model.predict(X_test)
print('  [5/6] XGBoost               done')

# 6. LightGBM
lgb_model = lgb.LGBMRegressor(
    n_estimators=300, learning_rate=0.05, max_depth=7,
    num_leaves=50, subsample=0.8, colsample_bytree=0.8,
    random_state=42, verbose=-1
)
lgb_model.fit(X_train, y_train, eval_set=[(X_val, y_val)],
              callbacks=[lgb.early_stopping(50, verbose=False),
                         lgb.log_evaluation(period=-1)])
y_pred_lgb = lgb_model.predict(X_test)
print('  [6/6] LightGBM              done')

pred_map = {
    'Baseline (Mean)'  : y_pred_dummy,
    'Linear Regression': y_pred_lr,
    'Ridge Regression' : y_pred_ridge,
    'Random Forest'    : y_pred_rf,
    'XGBoost'          : y_pred_xgb,
    'LightGBM'         : y_pred_lgb,
}

rows = []
for name, preds in pred_map.items():
    rmse = float(np.sqrt(mean_squared_error(y_test, preds)))
    mae  = float(mean_absolute_error(y_test, preds))
    r2   = float(r2_score(y_test, preds))
    rows.append({'Model': name, 'RMSE': round(rmse,4),
                 'MAE': round(mae,4), 'R2': round(r2,4)})

results_df = pd.DataFrame(rows).sort_values('RMSE').reset_index(drop=True)
results_df.to_csv('figures/table1_model_comparison.csv', index=False)

# Separate ordered list without baseline for per-model plots
all_models = [m for m in results_df['Model'].tolist() if m != 'Baseline (Mean)']
best_name  = all_models[0]
best_pred  = pred_map[best_name]

print('\n' + '='*58)
print('TABLE 2 — Model Performance on Test Set')
print('='*58)
print(results_df.to_string(index=False))
print('='*58)
print(f'\nBest non-baseline model: {best_name}')
```

```python
# Use only original 13 non-dummy feature names for display readability
DISPLAY_FEATS = ['danceability','energy','loudness','mode','speechiness',
                 'acousticness','instrumentalness','liveness','valence',
                 'tempo','time_signature','duration_s','explicit']

# Aggregate key dummy importances back into a single 'key' entry
rf_imp_full  = pd.Series(rf.feature_importances_,  index=FEATURE_COLS)
xgb_imp_full = pd.Series(xgb_model.feature_importances_, index=FEATURE_COLS)

rf_imp  = rf_imp_full[DISPLAY_FEATS].copy()
xgb_imp = xgb_imp_full[DISPLAY_FEATS].copy()
# Add aggregated key importance
rf_imp['key (agg.)']  = rf_imp_full[KEY_DUMMIES].sum()
xgb_imp['key (agg.)'] = xgb_imp_full[KEY_DUMMIES].sum()

combined = pd.concat([rf_imp.rename('RF'), xgb_imp.rename('XGB')], axis=1)
combined['RF_n']  = combined['RF']  / combined['RF'].max()
combined['XGB_n'] = combined['XGB'] / combined['XGB'].max()
combined['avg']   = (combined['RF_n'] + combined['XGB_n']) / 2
combined = combined.sort_values('avg', ascending=True)

fig, axes = plt.subplots(1, 3, figsize=(18, 7))

ax = axes[0]
norm_rf = plt.Normalize(combined['RF'].min(), combined['RF'].max())
rf_cols = [plt.cm.Blues(norm_rf(v)*0.7+0.3) for v in combined['RF']]
ax.barh(combined.index, combined['RF'], color=rf_cols, edgecolor='white', height=0.65)
ax.set_title('A  Random Forest\n(Mean Decrease Impurity)', fontsize=11)
ax.set_xlabel('Importance Score')
ax.axvline(combined['RF'].mean(), color='red', ls='--', lw=1.5,
           label=f'Mean = {combined["RF"].mean():.4f}')
ax.legend(fontsize=8)

ax2 = axes[1]
norm_xg = plt.Normalize(combined['XGB'].min(), combined['XGB'].max())
xg_cols = [plt.cm.Oranges(norm_xg(v)*0.7+0.3) for v in combined['XGB']]
ax2.barh(combined.index, combined['XGB'], color=xg_cols, edgecolor='white', height=0.65)
ax2.set_title('B  XGBoost\n(Gain)', fontsize=11)
ax2.set_xlabel('Importance Score')
ax2.axvline(combined['XGB'].mean(), color='red', ls='--', lw=1.5,
            label=f'Mean = {combined["XGB"].mean():.4f}')
ax2.legend(fontsize=8)

ax3 = axes[2]
y_pos = np.arange(len(combined))
ax3.barh(y_pos-0.2, combined['RF_n'],  height=0.38,
         color=PAL_ORANGE, alpha=0.8, label='Random Forest', edgecolor='white')
ax3.barh(y_pos+0.2, combined['XGB_n'], height=0.38,
         color=PAL_RED,    alpha=0.8, label='XGBoost',       edgecolor='white')
ax3.set_yticks(y_pos)
ax3.set_yticklabels(combined.index, fontsize=9)
ax3.set_xlabel('Normalised Importance (0–1)')
ax3.set_title('C  Head-to-Head\n(Normalised)', fontsize=11)
ax3.legend()

fig.suptitle('Figure 8 — Feature Importance: Random Forest vs. XGBoost',
             fontsize=14, fontweight='bold', y=1.02)
plt.tight_layout()
plt.savefig('figures/VIZ8_feature_importance.png')
plt.show()
print('VIZ 8 saved.')
```

```python
print('Computing learning curve for XGBoost (~1–2 min)...')

train_sizes, train_scores, val_scores = learning_curve(
    xgb.XGBRegressor(n_estimators=100, learning_rate=0.1,
                     max_depth=6, random_state=42, verbosity=0),
    X_train, y_train,
    train_sizes=np.linspace(0.1, 1.0, 9),
    cv=3,
    scoring='neg_root_mean_squared_error',
    n_jobs=-1
)

train_rmse = -train_scores
val_rmse   = -val_scores

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

ax = axes[0]
ax.fill_between(train_sizes,
                train_rmse.mean(1) - train_rmse.std(1),
                train_rmse.mean(1) + train_rmse.std(1),
                alpha=0.15, color=PAL_BLUE)
ax.fill_between(train_sizes,
                val_rmse.mean(1) - val_rmse.std(1),
                val_rmse.mean(1) + val_rmse.std(1),
                alpha=0.15, color=PAL_ORANGE)
ax.plot(train_sizes, train_rmse.mean(1), 'o-', color=PAL_BLUE,
        lw=2.2, label='Training RMSE')
ax.plot(train_sizes, val_rmse.mean(1),   's-', color=PAL_ORANGE,
        lw=2.2, label='Validation RMSE (CV-3)')
ax.set_xlabel('Training Set Size')
ax.set_ylabel('RMSE')
ax.set_title('A  Learning Curve — XGBoost')
ax.legend()

ax2 = axes[1]
gap = val_rmse.mean(1) - train_rmse.mean(1)
ax2.fill_between(train_sizes, 0, gap, alpha=0.25, color=PAL_RED)
ax2.plot(train_sizes, gap, 'o-', color=PAL_RED, lw=2.2,
         label='Generalisation Gap')
ax2.axhline(0, color='black', lw=1, ls='--')
ax2.set_xlabel('Training Set Size')
ax2.set_ylabel('RMSE Difference (Val – Train)')
ax2.set_title('B  Generalisation Gap\n(closer to 0 = less overfitting)')
ax2.legend()

fig.suptitle('Figure 11 — Learning Curve: XGBoost',
             fontsize=14, fontweight='bold', y=1.02)
plt.tight_layout()
plt.savefig('figures/VIZ11_learning_curve.png')
plt.show()
print('VIZ 11 saved.')
```
